(5 поена)

Neka su $A$ i $B$ pozitivno definitne matrice.

Po definiciji pozitivne definitnosti važi

$$
x^T A x > 0 \quad \text{za svako } x \neq 0
$$

$$
y^T B y > 0 \quad \text{za svako } y \neq 0
$$

Koristimo kombinaciju osobina Kronekerovog proizvoda (*Osobine 4 i 7* u okviru *Teoreme 2* u radnoj svesci *31BlokMatrice*):

$$
(x \otimes y)^T (A \otimes B) (x \otimes y) = (x^T A x)(y^T B y)
$$

gde je $x \otimes y$ Kronekerov proizvod vektora,  a na desnoj strani je obično množenje jer su u pitanju skalari.

Pošto važi

$$
x^T A x > 0
$$

i

$$
y^T B y > 0
$$

Sledi da je

$$
(x^T A x)(y^T B y) > 0
$$

za svaki nenulti vektor $x \otimes y$.

Odatle sledi da je

$$
A \otimes B
$$

pozitivno definitna matrica.

**Drugi način**

Po definiciji, simetrična matrica čije su sve sopstvene vrednosti strogo pozitivne je pozitivno definitna.

Prema pravilu iz sveske, sopstvene vrednosti Kronekerovog proizvoda $A \otimes B$ jednake su proizvodima sopstvenih vrednosti matrica koje ga čine, odnosno $\lambda_i \mu_j$.

S obzirom na to da su $\lambda_i > 0$ i $\mu_j > 0$, njihov proizvod je uvek strogo pozitivan, odnosno $\lambda_i \mu_j > 0$.

Pošto je matrica $A \otimes B$ simetrična i ima isključivo pozitivne sopstvene vrednosti, zaključujemo da je i ona pozitivno definitna matrica.

г) Показати да је Кронекеров производ две матрице ортогоналне пројекције поново матрица ортогоналне пројекције. Шта је простро колона ове матрице пројекције? Направити нумерички експеримент овог тврђења.


(5 поена)

Po definiciji, matrica je matrica ortogonalne projekcije ako i samo ako je simetrična ($P^T = P$) i idempotentna ($P^2 = P$). 

Neka su $A$ i $B$ matrice ortogonalne projekcije, što znači da važi $A^T = A$, $A^2 = A$, $B^T = B$ i $B^2 = B$.

Za Kronekerov proizvod $A \otimes B$ moramo dokazati ista dva svojstva:

- Simetričnost: Prema *osobini 4* iz *Teoreme 2*, transponovanje Kronekerovog proizvoda daje $(A \otimes B)^T = A^T \otimes B^T$. Kako su $A$ i $B$ simetrične, sledi $(A \otimes B)^T = A \otimes B$.

- Idempotentnost: Prema *osobini 7* iz *Teoreme 2*, važi $(A \otimes B)(C \otimes D) = (AC) \otimes (BD)$. Ako ovo primenimo na kvadrat matrice $A \otimes B$, dobijamo $(A \otimes B)^2 = (A \otimes B)(A \otimes B) = (A^2 \otimes B^2)$. Kako su $A$ i $B$ idempotentne, sledi $(A^2 \otimes B^2) = A \otimes B$.

Pošto je matrica $A \otimes B$ i simetrična i idempotentna, ona predstavlja matricu ortogonalne projekcije.

Prostor kolona matrice ortogonalne projekcije $A \otimes B$ predstavlja Kronekerov proizvod prostora kolona matrice $A$ i prostora kolona matrice $B$. Ako matrica $A$ projektuje vektore na potprostor $U$, a matrica $B$ projektuje na potprostor $V$, tada matrica $A \otimes B$ projektuje na potprostor $U \otimes V$.

````python
X1 = np.random.rand(4, 2)
Q1, _ = np.linalg.qr(X1)
A = Q1 @ Q1.T
print("A^2 == A:", np.allclose(A @ A, A))
print("A^T == A:", np.allclose(A.T, A))

X2 = np.random.rand(3, 1)
Q2, _ = np.linalg.qr(X2)
B = Q2 @ Q2.T
print("B^2 == B:", np.allclose(B @ B, B))
print("B^T == B:", np.allclose(B.T, B))

C = np.kron(A, B)

print("C^2 == C:", np.allclose(C @ C, C))
print("C^T == C:", np.allclose(C, C.T))
````

а) Показати да је за Фробенијусову норму важи $\|A\otimes B\|_F=\|A\|_F\|B\|_F.$ Направити нумерички експеримент овог тврђења.

(5 поена)

Prema definiciji, kvadrat Frobenijusove norme matrice $X$ jednak je tragu matrice $X^T X$, odnosno $\|X\|_F^2 = {\rm tr}(X^T X)$.

Ako primenimo ovu definiciju na Kronekerov proizvod, dobijamo $\|A \otimes B\|_F^2 = {\rm tr}((A \otimes B)^T (A \otimes B))$.

Na osnovu *Osobine 4* iz *Teoreme 2*, transponovanje Kronekerovog proizvoda daje $(A \otimes B)^T = A^T \otimes B^T$.

Dalje, primenom *Osobine 7* iz *Teoreme 2*, izraz pod tragom postaje $(A^T \otimes B^T)(A \otimes B) = (A^T A) \otimes (B^T B)$.

Sada izraz za kvadrat norme izgleda ovako: $\|A \otimes B\|_F^2 = {\rm tr}((A^T A) \otimes (B^T B))$.

Prema *Teoremi 1* iz sveske *31BlokMatrice* dobijamo ${\rm tr}((A^T A) \otimes (B^T B)) = {\rm tr}(A^T A){\rm tr}(B^T B)$.

Kako je ${\rm tr}(A^T A) = \|A\|_F^2$ i ${\rm tr}(B^T B) = \|B\|_F^2$, sledi da je $\|A \otimes B\|_F^2 = \|A\|_F^2 \|B\|_F^2$.

Korenovanjem obe strane dolazimo do konačnog izraza: $\|A \otimes B\|_F = \|A\|_F \|B\|_F$.

````python
A = np.random.rand(4, 3)
B = np.random.rand(2, 5)

C = np.kron(A, B)

norm_A = np.linalg.norm(A, 'fro')
norm_B = np.linalg.norm(B, 'fro')
norm_C = np.linalg.norm(C, 'fro')

print(np.isclose(norm_C, norm_A * norm_B))
````

б) Показати да је за индуковане норму важи $\|A\otimes B\|_p=\|A\|_p\|B\|_p.$ Направити нумерички експеримент овог тврђења.

(5 поена)

Prema definiciji, indukovana $p$-norma matrice $X$ definisana je preko $p$-norme vektora kao $\|X\|_p = \max_{v \neq 0} \frac{\|Xv\|_p}{\|v\|_p}$.

Najpre, za Kronekerov proizvod dva vektora $x$ i $y$, može se pokazati da važi jednakost za $p$-norme: $\|x \otimes y\|_p = \|x\|_p \|y\|_p$.

$$ \|x\|_p = \left( \sum_{i=1}^m |x_i|^p \right)^{1/p}, \quad
   \|y\|_p = \left( \sum_{j=1}^n |y_j|^p \right)^{1/p} $$

$$
x \otimes y =
\begin{bmatrix}
x_1 y \\
x_2 y \\
\vdots \\
x_m y
\end{bmatrix}, 
\quad (x \otimes y)_{ij} = x_i y_j
$$

$$
\|x \otimes y\|_p 
= \left( \sum_{i=1}^m \sum_{j=1}^n |x_i y_j|^p \right)^{1/p}
= \left( \sum_{i=1}^m |x_i|^p \sum_{j=1}^n |y_j|^p \right)^{1/p}
$$

$$
= \left( \sum_{i=1}^m |x_i|^p \right)^{1/p} 
   \left( \sum_{j=1}^n |y_j|^p \right)^{1/p}
= \|x\|_p \, \|y\|_p
$$

Prema *osobini 7* iz *Teoreme 2*, množenje Kronekerovog proizvoda matrica $A \otimes B$ vektorom oblika $x \otimes y$ daje:

$$(A \otimes B)(x \otimes y) = (Ax) \otimes (By)$$

Ako primenimo $p$-normu na ovaj izraz, dobijamo:

$$\|(A \otimes B)(x \otimes y)\|_p = \|(Ax) \otimes (By)\|_p = \|Ax\|_p \|By\|_p$$

Pošto indukovana norma predstavlja maksimalno uvećanje dužine vektora, maksimum po svim mogućim vektorima prostora $v$ mora biti veći ili jednak maksimumu po vektorima koji se mogu zapisati kao Kronekerov proizvod $x \otimes y$. 

Zato važi donja granica:
$$\|A \otimes B\|_p \geq \max_{x \neq 0, y \neq 0} \frac{\|(A \otimes B)(x \otimes y)\|_p}{\|x \otimes y\|_p} = \max_{x \neq 0} \frac{\|Ax\|_p}{\|x\|_p} \cdot \max_{y \neq 0} \frac{\|By\|_p}{\|y\|_p} = \|A\|_p \|B\|_p$$

Može se pokazati da važi i gornja granica $\|A \otimes B\|_p \leq \|A\|_p \|B\|_p$, iz čega direktno sledi tražena jednakost:

$$\|A \otimes B\|_p = \|A\|_p \|B\|_p$$

````python
A = np.random.rand(4, 3)
B = np.random.rand(2, 2)

C = np.kron(A, B)

p = 2
norm_A = np.linalg.norm(A, ord=p)
norm_B = np.linalg.norm(B, ord=p)
norm_C = np.linalg.norm(C, ord=p)

norm_product = norm_A * norm_B

print(f"Indukovana p-norma Kronekerovog proizvoda: {norm_C}")
print(f"Proizvod indukovanih p-normi A i B: {norm_product}")
print(f"Tvrđenje je tačno: {np.isclose(norm_C, norm_product)}")
````

**Задатак 4.** а) За Кронекерову суму матрица $A\in\mathcal{M}_{m\times m}$ и $B\in\mathcal{M}_{n\times n},$ показати једнакост:
$${\rm tr}(A\oplus B)=n\,{\rm tr}(A)+m\,{\rm tr}(B).$$


(5 поена)

Prema definiciji iz sveske *31BlokMatrice*, Kronekerova suma dve kvadratne matrice definisana je kao:

$$A \oplus B = A \otimes I_n + I_m \otimes B$$

gde je $I_n$ jedinična matrica dimenzija $n \times n$, a $I_m$ jedinična matrica dimenzija $m \times m$.

Ako primenimo operaciju traga na ceo izraz, dobijamo:

$${\rm tr}(A \oplus B) = {\rm tr}(A \otimes I_n + I_m \otimes B)$$

Kako je trag linearni operator, trag zbira dve matrice jednak je zbiru njihovih tragova (${\rm tr}(X + Y) = {\rm tr}(X) + {\rm tr}(Y)$):

$${\rm tr}(A \otimes I_n + I_m \otimes B) = {\rm tr}(A \otimes I_n) + {\rm tr}(I_m \otimes B)$$

Prema *Teoremi 1* iz sveske, trag Kronekerovog proizvoda dve kvadratne matrice jednak je proizvodu njihovih tragova, odnosno ${\rm tr}(X \otimes Y) = {\rm tr}(X)\,{\rm tr}(Y)$. Primenom ove teoreme dobijamo:

$${\rm tr}(A \otimes I_n) + {\rm tr}(I_m \otimes B) = {\rm tr}(A)\,{\rm tr}(I_n) + {\rm tr}(I_m)\,{\rm tr}(B)$$

Znamo da je trag jedinične matrice jednak njenoj dimenziji (jer se na dijagonali nalaze samo jedinice čiji je broj jednak dimenziji matrice). Zato važi:

$${\rm tr}(I_n) = n$$

$${\rm tr}(I_m) = m$$

Zamenom ovih vrednosti dolazimo do traženog rešenja:

$${\rm tr}(A)\,{\rm tr}(I_n) + {\rm tr}(I_m)\,{\rm tr}(B) = {\rm tr}(A) \cdot n + m \cdot {\rm tr}(B)$$

odnosno:

$${\rm tr}(A \oplus B) = n\,{\rm tr}(A) + m\,{\rm tr}(B)$$

б) Пронаћи израз за $\det(A\oplus B)$ у зависности од сопствених вредности матрица $A\in\mathcal{M}_{m\times m}$ и $B\in\mathcal{M}_{n\times n}.$

(5 поена)

Determinanta svake kvadratne matrice jednaka je proizvodu svih njenih sopstvenih vrednosti.

Neka su $\lambda_1, \lambda_2, \dots, \lambda_m$ sopstvene vrednosti matrice $A$.

Neka su $\mu_1, \mu_2, \dots, \mu_n$ sopstvene vrednosti matrice $B$.

Prema pravilu iz *Primera 14* iz sveske, sopstvene vrednosti Kronekerove sume $A \oplus B$ jednake su zbiru svih mogućih kombinacija sopstvenih vrednosti matrica $A$ i $B$.

Dakle, sopstvene vrednosti matrice $A \oplus B$ su $\lambda_i + \mu_j$ za svako $i \in \{1, \dots, m\}$ i $j \in \{1, \dots, n\}$.

Množenjem svih ovih sopstvenih vrednosti dolazimo do konačnog izraza za determinantu:

$$\det(A \oplus B) = \prod_{i=1}^{m} \prod_{j=1}^{n} (\lambda_i + \mu_j)$$

**Задатак 5.** За наредна тврђења могу се користити резултати доказани у радној свесци 31BlokMatrice.

а) Доказати важан идентитет који повезује Катри–Рао производ ($\odot$) и Адамаров (Hadamard) производ ($\ast$):
$$
(A \odot B)^{T}(A \odot B) = (A^{T}A) \ast (B^{T}B),
$$
где су $A \in M_{m \times n}$ и $B \in M_{p \times n}$ матрице са истим бројем колона.

**Сугестија:** Посматрати скаларни производ колона Катри–Рао производа  
$(a_i \otimes b_i)^{T}(a_j \otimes b_j)$  
и искористити особине Кронекеровог производа.

(10 поена)

Neka je matrica $C = A \odot B$. 

Po definiciji Katri-Rao proizvoda, $i$-ta kolona matrice $C$ predstavlja Kronekerov proizvod odgovarajućih kolona matrica $A$ i $B$, odnosno $c_i = a_i \otimes b_i$.

Element na poziciji $(i, j)$ matrice $(A \odot B)^{T}(A \odot B)$ je skalarni proizvod $i$-te i $j$-te kolone matrice $C$:

$$[(A \odot B)^{T}(A \odot B)]_{ij} = c_i^{T} c_j = (a_i \otimes b_i)^{T} (a_j \otimes b_j)$$

Primenom osobine transponovanja Kronekerovog proizvoda dobija se:

$$(a_i \otimes b_i)^{T} = a_i^{T} \otimes b_i^{T}$$

Primenom osobine mešovitog proizvoda za Kronekerov proizvod sledi:

$$(a_i^{T} \otimes b_i^{T})(a_j \otimes b_j) = (a_i^{T} a_j) \otimes (b_i^{T} b_j)$$

Pošto su izrazi $a_i^{T} a_j$ i $b_i^{T} b_j$ zapravo skalari (rezultati skalarnog množenja vektora), njihov Kronekerov proizvod je ekvivalentan običnom množenju skalara:

$$(a_i^{T} a_j) \otimes (b_i^{T} b_j) = (a_i^{T} a_j)(b_i^{T} b_j)$$

Skalarni proizvod $a_i^{T} a_j$ predstavlja tačno element na poziciji $(i, j)$ matrice $A^{T}A$, dok je $b_i^{T} b_j$ element na poziciji $(i, j)$ matrice $B^{T}B$:

$$a_i^{T} a_j = [A^{T}A]_{ij}, \quad b_i^{T} b_j = [B^{T}B]_{ij}$$

Zamenom ovih vrednosti, zaključujemo da je $(i, j)$-ti element rezultujuće matrice jednak proizvodu $(i, j)$-tih elemenata matrica $A^{T}A$ i $B^{T}B$:

$$[(A \odot B)^{T}(A \odot B)]_{ij} = [A^{T}A]_{ij} \cdot [B^{T}B]_{ij}$$

Množenje matrica element po element po definiciji predstavlja Adamarov proizvod, čime je dokazana jednakost:

$$(A \odot B)^{T}(A \odot B) = (A^{T}A) \ast (B^{T}B)$$

б) Направити нумерички експеримент који потврђује тврђење задатака под а).  
Измерити време потребно за израчунавање леве и десне стране једнакости за матрице већих димензија (нпр. $n = 500$) и прокоментарисати разлику у брзини.

(5 поена)

````python
n = 500
m = 100
p = 100

A = np.random.randn(m,n)
B = np.random.randn(p,n)

def khatri_rao(A,B):
    return np.vstack([np.kron(A[:,i],B[:,i]) for i in range(A.shape[1])]).T

start = timer()
KR = khatri_rao(A,B)
left = KR.T @ KR
time_left = timer() - start


start = timer()
right = (A.T @ A) * (B.T @ B)
time_right = timer() - start

print("Jednakost:", np.allclose(left, right))
print("Vreme leva strana:", time_left)
print("Vreme desna strana:", time_right)
````

Leva strana je sporija jer mora da izvrši Katri-Rao proizvod čime se dobija matrica dimenzije $mp \times n$ i zatim izvrši množenje velikih matrica.

Sa druge strane, desna strana računa samo
$$
(A^T A) \ast (B^T B),
$$
gde je $\ast$ Adamarov proizvod.

Ovde se najpre računaju matrice $A^T A$ i $B^T B$, koje imaju dimenzije $n \times n$, a zatim se primenjuje Adamarov proizvod nad njima.

Zbog toga je desna strana značajno brža i memorijski efikasnija, naročito za velike dimenzije kao što je $n = 500$.

в) Нека матрице $A \in M_{m \times n}$ и $B \in M_{p \times n}$ имају ортонормиране колоне  (тј. $A^{T}A = I$ и $B^{T}B = I$).  Доказати да тада и њихов Катри–Рао производ  
$$
C = A \odot B
$$
има ортонормиране колоне.

**Сугестија:** Искористити идентитет доказан под а).

(5 поена)

Da bismo pokazali da i matrica $C = A \odot B$ ima ortonormirane kolone, potrebno je dokazati da važi:

$$C^T C = I$$

Oslanjajući se na identitet koji je dokazan u delu pod $(a)$, imamo:

$$C^T C = (A \odot B)^T (A \odot B) = (A^T A) \ast (B^T B)$$

Zamenom uslova ortonormiranosti ($A^T A = I$ i $B^T B = I$) u desnu stranu jednakosti, dobijamo:

$$(A^T A) \ast (B^T B) = I \ast I$$

Adamarov proizvod predstavlja množenje matrica element po element. Jedinična matrica $I$ ima jedinice na glavnoj dijagonali i nule na svim ostalim pozicijama. Množenjem takve matrice sa samom sobom element po element, na dijagonali dobijamo $1 \cdot 1 = 1$, a na ostalim pozicijama $0 \cdot 0 = 0$. Prema tome, važi:

$$I \ast I = I$$

Iz ovoga direktno sledi konačan zaključak:

$$C^T C = I$$

što znači da kolone matrice $C$ obrazuju ortonormiran sistem.

г) Ако су $P_A = AA^{T}$ и $P_B = BB^{T}$ матрице ортогоналне пројекције на просторе колона матрица $A$ и $B$,  објаснити зашто је израчунавање матрице пројекције на простор колона Катри–Рао производа $A \odot B$
рачунски много једноставније када су колоне ортонормиране.
Направити нумерички експеримент који то потврђује.

(5 поена)

Neka je
$$
C = A \odot B
$$

Projekcija na prostor kolona matrice $C$ u opštem slučaju je
$$
P_C = C (C^T C)^{-1} C^T
$$

Međutim, ako matrice $A$ i $B$ imaju ortonormirane kolone, tada važi
$$
A^T A = I, \qquad B^T B = I
$$

Tada i njihov Katri-Rao proizvod ima ortonormirane kolone, odnosno
$$
C^T C = I
$$

U tom slučaju formula za projekciju se značajno pojednostavljuje i postaje
$$
P_C = C C^T.
$$

Razlog zašto je izračunavanje brže je taj što u opštem slučaju moramo da računamo matricu $C^T C$ i njen inverz, što je računski skupo.

Kada su kolone ortonormirane, važi $C^T C = I$, pa računanje inverza nije potrebno. 
Zbog toga se projekcija svodi na jednostavan proizvod
$$
P_C = C C^T
$$
što značajno ubrzava izračunavanje.

````python
m = 100
p = 100
n = 700

A = np.random.randn(m,n)
B = np.random.randn(p,n)

A,_ = np.linalg.qr(A)
B,_ = np.linalg.qr(B)

C = khatri_rao(A,B)

start = timer()
P_general = C @ np.linalg.inv(C.T @ C) @ C.T
time1 = timer() - start

start = timer()
P_fast = C @ C.T
time2 = timer() - start

print("Razlika projekcija:", np.linalg.norm(P_general - P_fast))
print("Vreme opšta formula:", time1)
print("Vreme ortonormirane kolone:", time2)
````

б) Реализовати функцију у Python-у која решава овај проблем **без експлицитног формирања Катри–Рао производа**.

**Помоћ:** За ефикасно израчунавање десне стране
$$
z = (A \odot B)^{T} y,
$$
приметити да ако вектор $y$ реструктурирамо у матрицу $Y \in M_{p \times m}$ (тако да је $y = vec(Y)$),  
тада је $i$-ти елемент вектора $z$ једнак
$$
z_i = b_i^{T} \, Y \, a_i,
$$
где су $a_i$ и $b_i$ колоне матрица $A$ и $B$.

У NumPy-ју се овај вектор може добити **без петљи**, коришћењем матричног множења и сумирања дуж одређене осе, или преко:
```python
np.einsum('pi,pm,mi->i', B, Y, A)
```
